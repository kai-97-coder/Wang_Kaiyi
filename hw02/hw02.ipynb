```python
exclude = ["US", "GU", "PR", "VI"]

binge = binge[~binge["location_abbr"].isin(exclude)].copy()
pvt = pvt[~pvt["location_abbr"].isin(exclude)].copy()
```

```python
binge_all = binge.loc[
    binge["group"] == "Overall",
    ["location_abbr", "location", "year", "prevalence"]
].rename(columns={"prevalence": "binge_all"})
#First select rows where group is overall, keep four columns, rename prevalence to bingge_all.
binge_all.head()
```

```python
binge_male = binge.loc[
    binge["group"] == "Male",
    ["location_abbr", "location", "year", "prevalence"]
].rename(columns={"prevalence": "binge_male"})

binge_male.head()
```

```python
binge_female = binge.loc[
    binge["group"] == "Female",
    ["location_abbr", "location", "year", "prevalence"]
].rename(columns={"prevalence": "binge_female"})

binge_female.head()
```

```python
binge_combined = pd.merge(
    binge_all,
    binge_male,
    on=["location_abbr", "location", "year"],
    how="left"
)
#according to keys, merge tables.
binge_combined.head()
```

```python
binge_wide = pd.merge(
    binge_combined,
    binge_female,
    on=["location_abbr", "location", "year"],
    how="left"
)
#same thing, assign binge_wide since it's side now.
binge_wide.head()
```

## Question 3: Poverty and binge drinking

```python
comparison = state_year[
    state_year["year"].isin([2019, 2021])
    & (state_year["state_abbr"] != "DC")
].dropna(subset=["binge_all", "poverty"]).copy()
# select either year (2019 2021), exclude DC, drop missing values (either is missing)
comparison.groupby("year")["state"].count()
#count the eligible states separately for each year
```

```python
comparison.groupby("year")["poverty"].median()
```

```python
comparison["poverty_median"] = (
    comparison.groupby("year")["poverty"].transform("median")
)
# give one result per original row
comparison[["state", "year", "poverty", "poverty_median"]].head()
```

```python
comparison["poverty_group"] = "Below median"
# new column poverty_group, all below median
comparison.loc[
    comparison["poverty"] >= comparison["poverty_median"],
    "poverty_group"
] = "At or above median"
# change to at or above after compare
comparison[
    ["state", "year", "poverty", "poverty_median", "poverty_group"]
].head()
```

```python
poverty_summary = (
    comparison
    .groupby(["year", "poverty_group"])["binge_all"]
    .agg(
        number_of_states="count",
        average_binge_rate="mean"
    )
    .reset_index()
)
#create group for year and poverty group, select overall binge level, count and calculate mean, rest index.
poverty_summary
```

## Question 3C
After dropping missing values and excluding DC, the at-or-above-median poverty group had a slightly lower average binge drinking rate in 2019 (12.23%) than the below-median group (12.35%). In 2021, the pattern reversed. The at-or-above-median group drinks a little more (11.25%) than the below-median group (10.65%). Hence, higher drinking was associated with higher poverty only in 2021.
