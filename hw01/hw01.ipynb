## 4. Non-linear function and plot

```python
import matplotlib.pyplot as plt #load the plotting tools with the short name plt

x = list(range(1001)) #generate x from 0-1000
y = [value ** 2 for value in x] #generate y

plt.plot(x, y) #Plotting and explanation
plt.xlabel("x")
plt.ylabel("y")
plt.title("Quadratic function: y = x^2")
plt.show()
```

Explanation: The function plotted above is:

$$
y = x^2
$$

Each y-value is the square of its corresponding x-value.
