### 1a

```python
def session_cost(
    hours: float,
    exam: str = "GRE",
    online: bool = False
) -> float:
    """Return the session cost before package discounts."""
    rates = {"GRE": 90.0, "GMAT": 110.0, "LSAT": 130.0}
    total = hours * rates[exam]

    if online:
        total = total * 0.85

    return total
```

```python
session_cost(10, exam="LSAT", online=False)
```

### 1b

```python
def package_discount(
    cost: float,
    hours: float,
    promo_code: str | None = None
) -> float:
    """Return the discount amount for a tutoring package."""
    if hours < 10:
        discount = 0.0
    elif hours < 20:
        discount = cost * 0.10
    else:
        discount = cost * 0.15

    if promo_code == "COLUMBIA":
        discount = discount + 50.0

    return discount
```

```python
print(package_discount(1300, 10))
print(package_discount(2000, 20, promo_code="COLUMBIA"))
```

## 1c

```python
# Print a tutoring quote and return the cost after discounts.
def quote_calculator(hours: float, **kwargs) -> float:
    exam = kwargs.get("exam", "GRE")
    online = kwargs.get("online", False)
    promo_code = kwargs.get("promo_code", None)

    if hours <= 0:
        raise ValueError(
            f"Hours must be positive; received {hours}."
        )

    if exam not in ["GRE", "GMAT", "LSAT"]:
        raise ValueError(
            f"Exam must be GRE, GMAT, or LSAT; received {exam}."
        )

    cost = session_cost(hours, exam=exam, online=online)
    discount = package_discount(cost, hours, promo_code=promo_code)
    total = cost - discount

    if online:
        session_format = "online"
    else:
        session_format = "in person"

    rate_per_hour = cost / hours

    print(f"Exam: {exam}")
    print(f"Hours booked: {hours}")
    print(f"Format: {session_format}")
    print(f"Rate per hour: ${rate_per_hour:.2f}")
    print("---------------------------")
    print(f"Session cost: ${cost:.2f}")
    print(f"Package discount: ${discount:.2f}")
    print("---------------------------")
    print(f"Total cost: ${total:.2f}")

    return total
```

```python
quote = quote_calculator(
    24,
    exam="GMAT",
    online=True,
    promo_code="COLUMBIA"
)
```

## 3

Before running the code, I predict spring roster will contain only Ana and fall roster will contain both Ana and Ben.

```python
def book_student(roster, name, hours):
    roster.append({"name": name, "hours": hours})
    return roster

spring_roster = [{"name": "Ana", "hours": 10}]
fall_roster = book_student(spring_roster, "Ben", 12)

print(spring_roster)
print(fall_roster)
```

This happens because both name refer to the same list, in other word, passing a list into a funciton does not automatically copy it.

```python
spring_roster is fall_roster
```

```python
def book_student_pure(roster, name, hours):
    new_roster = roster.copy()
    new_roster.append({"name": name, "hours": hours})
    return new_roster

spring_roster = [{"name": "Ana", "hours": 10}]
fall_roster = book_student_pure(spring_roster, "Ben", 12)

print(spring_roster)
print(fall_roster)
```

```python
spring_roster is fall_roster
```

- My original prediction was wrong, since append() changed the original list.
- The original function return the same list and both refer to it.
- The new function copies the list, and keeps the original list unaffected.
